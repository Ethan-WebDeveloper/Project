## 1. Output, Variables, and Basic Data Types

A **variable** is a name that refers to a value in memory. Python determines the type dynamically when a value is assigned. Common built-in types include integers (`int`), decimal numbers (`float`), text (`str`), and Boolean values (`bool`). The `type()` function reveals the current type of a value.

```python
student = "Seojun Hwang"
age = 16
height = 175.5
is_learning = True

print("Student:", student)
print(type(age), type(height), type(student), type(is_learning))
```

## 2. Operators and String Formatting

Arithmetic operators perform calculations. `/` produces a floating-point result, `//` performs floor division, and `%` returns the remainder. An f-string places expressions inside `{ }`, making output easier to read and format.

```python
a, b = 13, 5
print("Addition:", a + b)
print("Division:", a / b)
print("Quotient and remainder:", a // b, a % b)
print(f"{student} studied Python today.")
```

## 3. Lists and Dictionaries

A **list** stores an ordered, mutable sequence and uses zero-based indexes. A **dictionary** stores key-value pairs and retrieves values by meaningful keys. These structures are widely used to organize records before converting them into pandas data structures.

```python
scores = [85, 92, 78, 96]
student_info = {"name": "Seojun Hwang", "course": "Python", "completed": True}

print("First score:", scores[0])
print("Average:", sum(scores) / len(scores))
print("Course:", student_info["course"])

scores.append(100)
print("After append:", scores)
```

## 4. Conditional Statements

Conditional statements choose which block of code to run. Python checks `if`, then any `elif` branches, and finally `else`. Indentation defines the code belonging to each branch, so consistent spacing is part of Python syntax.

```python
score = 88

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "C"

print(f"The grade for a score of {score} is {grade}.")
```

## 5. Loops

A `for` loop processes each item in a collection. `enumerate()` supplies both the position and value, while `range()` creates a sequence of integers. Accumulator variables such as `total` are often used to combine results across iterations.

```python
stocks = ["Samsung Electronics", "SK Hynix", "Apple"]
for index, stock in enumerate(stocks, start=1):
    print(index, stock)

# Sum of even numbers from 1 through 10
total = 0
for number in range(1, 11):
    if number % 2 == 0:
        total += number
print("Sum of even numbers:", total)
```

## 6. Functions

A function packages reusable logic behind a clear name. Parameters receive input values, and `return` sends a result back to the caller. A docstring documents the purpose of the function and can be viewed with `help()`.

```python
def calculate_return(start_price, end_price):
    """Calculate percentage return from a starting and ending price."""
    return (end_price / start_price - 1) * 100

result = calculate_return(70000, 73500)
print(f"Return: {result:.2f}%")
```

## 7. Exception Handling

Exceptions are runtime errors that can be handled without abruptly stopping the program. A `try` block contains code that may fail, while specific `except` blocks explain how to respond. Catching specific exception types is safer than hiding every error.

```python
try:
    user_value = "10"
    number = int(user_value)
    print(100 / number)
except ValueError:
    print("The value cannot be converted to a number.")
except ZeroDivisionError:
    print("Division by zero is not allowed.")
```
